# <传感器名称> 适配器使用示例

本 Notebook 演示如何使用 TLabel 适配器将 <传感器名称> 数据转换为统一的 tlabel_v2 格式。

```python
# 安装 tlabel（如未安装）
# !pip install tlabel

from adapter.my_sensor import MySensorAdapter
from tlabel.export.writer import Exporter
```

## 1. 加载数据

```python
adapter = MySensorAdapter()

# TODO: 替换为你的数据路径
data = adapter.load('data/sample/')

print(f'传感器: {data.sensor_id}')
print(f'帧数: {data.num_frames}')
print(f'时长: {data.duration:.2f}s')
```

## 2. 查看 22 维特征

```python
import pandas as pd

# 将前几帧的特征转为 DataFrame 查看
rows = [f.tlabel_v2 for f in data.frames[:10]]
df = pd.DataFrame(rows)
print(df.to_string())
```

## 3. 查看传感器能力声明

```python
caps = adapter.get_capabilities()
supported = [k for k, v in caps.items() if v]
unsupported = [k for k, v in caps.items() if not v]

print(f'支持 ({len(supported)}/22):')
for dim in supported:
    print(f'  ✅ {dim}')
print(f'\n不支持 ({len(unsupported)}/22):')
for dim in unsupported:
    print(f'  ❌ {dim}')
```

## 4. 导出为其他格式

```python
exporter = Exporter()

# 导出为 JSON
exporter.to_json(data, 'output.json')
print('✅ 已导出 output.json')

# 导出为 CSV
exporter.to_csv(data, 'output.csv')
print('✅ 已导出 output.csv')
```
